## 1. Import required libraries

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries installed successfully!")

```

## 2. Load the CSV dataset

```python
import os

file_path = "data/student-mat.csv"

print("Current folder:", os.getcwd())

df = pd.read_csv(file_path, sep=";")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

df.head()
```

## 3. Inspect the dataset

```python
# View first and last five rows
display(df.head())
display(df.tail())

# Column names, data types and non-null counts
df.info()

# Statistical summary of numerical columns
df.describe()
```

## 4. Check missing values

```python
missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100

missing_report = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage.round(2)
})
display(missing_report)

print("Total missing values:", int(df.isnull().sum().sum()))
```

### Handle missing values

```python
# Separate numeric and categorical columns.
numeric_cols = df.select_dtypes(include=np.number).columns
categorical_cols = df.select_dtypes(exclude=np.number).columns

# Fill numeric missing values with the column median.
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill categorical missing values with the most frequent value.
for col in categorical_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after treatment:", int(df.isnull().sum().sum()))
```

## 5. Find and remove duplicate rows

```python
duplicate_count = int(df.duplicated().sum())
print("Duplicate rows found:", duplicate_count)

df = df.drop_duplicates().copy()
print("Shape after removing duplicates:", df.shape)
print("Remaining duplicate rows:", int(df.duplicated().sum()))
```

## 6. Calculate basic statistics

```python
# Mean, median, mode and standard deviation for numeric columns
statistics = pd.DataFrame({
    "Mean": df[numeric_cols].mean(),
    "Median": df[numeric_cols].median(),
    "Mode": df[numeric_cols].mode().iloc[0],
    "Standard Deviation": df[numeric_cols].std()
}).round(2)

display(statistics)

# Focus on final student grade (G3)
print("Average final grade (G3):", round(df["G3"].mean(), 2))
print("Median final grade (G3):", round(df["G3"].median(), 2))
```

## 7. Basic visualizations

### A. Bar chart – Average final grade by study time

```python
average_grade = df.groupby("studytime")["G3"].mean()

plt.figure(figsize=(8, 5))
average_grade.plot(kind="bar")
plt.title("Average Final Grade by Weekly Study Time")
plt.xlabel("Study Time Category (1 = less than 2 hours, 4 = more than 10 hours)")
plt.ylabel("Average Final Grade (G3)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("outputs/bar_average_grade_by_studytime.png", dpi=150)
plt.show()
```

### B. Histogram – Distribution of final grades

```python
plt.figure(figsize=(8, 5))
plt.hist(df["G3"], bins=10, edgecolor="black")
plt.title("Distribution of Students' Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.savefig("outputs/histogram_final_grades.png", dpi=150)
plt.show()
```

### C. Pie chart – Student count by gender

```python
gender_counts = df["sex"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(gender_counts, labels=gender_counts.index, autopct="%1.1f%%", startangle=90)
plt.title("Student Distribution by Gender")
plt.tight_layout()
plt.savefig("outputs/pie_gender_distribution.png", dpi=150)
plt.show()
```

## 8. Save the cleaned dataset

```python
cleaned_path = "data/student-mat-cleaned.csv"
df.to_csv(cleaned_path, index=False)
print("Cleaned dataset saved to:", cleaned_path)
```
